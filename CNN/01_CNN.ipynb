```python
import torch 
import torch.nn as nn

x = torch.randn([1,3,32,32])

conv = nn.Conv2d(
    in_channels=3,
    out_channels=16,
    kernel_size=3,
    stride=1,
    padding=1
)

output = conv(x)

print(output.shape)


#ReLU activation 

relu = nn.ReLU(x)

x = relu(x)

print(x.shape)

pool = nn.maxpool(kernal_size = 2,  stride = 2)

x = pool(x)

print(x)
```
