```python
"""
Testing first CNN
"""
import torch
import torch.nn as nn

###Classes scores from model
output = torch.tensor([0.2, 3.5, 1.1])

##True class--- Dog-1 
label = torch.tensor([1])

##Defining Loss Function
loss_fn = nn.CrossEntropyLoss()

##putting values in loss function
loss = loss_fn(output, label)

##Final Loss
print("Loss: ",loss)
```

```python
wrong_output = torch.tensor([3.5, 0.2, 1.1])

loss = loss_fn(wrong_output, label)

print('Loss: ',loss)
```
