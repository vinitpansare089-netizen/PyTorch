```python
"""
CNN Architecture
"""

import torch
import torch.nn as nn

class MyCNN(nn.Module):
    def __init__(self):
        super().__init__()

        self.features = nn.Sequential(
            nn.Conv2d(3, 8, kernel_size=3, stride=1, padding=1),
            nn.ReLU(),
            nn.MaxPool2d(kernel_size=2, stride=2),

            nn.Conv2d(8, 16, kernel_size=3, stride=1, padding=1),
            nn.ReLU(),
            nn.MaxPool2d(kernel_size=2, stride=2)

        )

        self.classifier = nn.Sequential(
            nn.Flatten(),
            nn.Linear(16 * 8 * 8 ,3)
        )

    def forward(self, x):
        x = self.features(x)
        x = self.classifier(x)
        return x

model = MyCNN()
print(model)
```

```python
"""
Creating a fake training example
"""
###IMAGE
x = torch.randn([1,3,32,32])

###Correct answer
y = torch.tensor([1])
```

```python
###forward pass

output = model(x)

print(output)
print(output.shape)

###Calculating loss
loss_fn = nn.CrossEntropyLoss()

loss = loss_fn(output, y)

print("Loss: ", loss.item())
```

```python
"""
After calculating loss optimizer
"""


optimizer = torch.optim.SGD(
    model.parameters(),
    lr = 0.01
)

output = model(x)
loss = loss_fn(output, y)

print("Before:", loss.item())

optimizer.zero_grad()
loss.backward()
optimizer.step()

output = model(x)
loss = loss_fn(output, y)

print("After:", loss.item())
```
