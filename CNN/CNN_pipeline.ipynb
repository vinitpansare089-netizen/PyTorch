```python
import torch 
import torch.nn as nn

x = torch.randn([1,3,32,32])

# print(x.shape)

conv = nn.Conv2d(
    in_channels = 3,
    out_channels = 8,
    kernel_size = 3,
    stride = 1,
    padding = 1
)

x = conv(x)
print(x.shape)##After passing from conv2D the x now shapes '''torch.Size([1, 8, 32, 32])'''

#####Now i am passing through ReLU functions

relu = nn.ReLU()

x = relu(x)

print(x.shape)

########POOLING of x

pool = nn.MaxPool2d(kernel_size=2,stride=2)

x = pool(x)
print(x.shape)#////torch.Size([1, 8, 16, 16])

#//////////////////////////////////////////////////

conv2 = nn.Conv2d(
    in_channels=8,
    out_channels=16,
    kernel_size=3,
    stride=1,
    padding=1
)

x = conv2(x)
print(x.shape)

###ReLU2
relu2 = nn.ReLU()

x = relu2(x)
print(x.shape)

####maxpool2

pool2 = nn.MaxPool2d(kernel_size=2,stride=2)

x = pool2(x)
print(x.shape)

###########FLATTEN #################

flatten = nn.Flatten()

x = flatten(x)
print(x.shape)

###Linear ########
```
