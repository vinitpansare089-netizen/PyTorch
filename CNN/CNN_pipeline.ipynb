```python
import torch 
import torch.nn as nn

x = torch.randn([1,3,32,32])

# print(x.shape)

conv = nn.Conv2d(
    in_channels = 3,
    out_channels = 8,
    kernel_size = 3,
    stride = 1,
    padding = 1
)

x = conv(x)
print(x.shape)##After passing from conv2D the x now shapes '''torch.Size([1, 8, 32, 32])'''

#####Now i am passing through ReLU functions

relu = nn.ReLU()

x = relu(x)

print(x.shape)

```
