```python
"""
Implementing CNNs
"""
import torch
import torch.nn as nn

x = torch.arange(75, dtype=torch.float32).reshape(1, 3, 5, 5)

# print(x)

"""Creating a filter below"""

conv = nn.Conv2d(
    in_channels=3,
    out_channels=1,
    padding=0,
    stride=1,
    kernel_size=3,
    bias=False
)

with torch.no_grad():
    conv.weight.fill_(1)

output = conv(x)

print(output)

print(x)

# print(conv.weight)
# print(conv.bias)
```
