```python
"""
Implementing CNNs
"""
import torch
import torch.nn as nn

x = torch.randn(1,3,5,5)

# print(x.shape)

"""Creating a filter below"""

conv = nn.Conv2d(
    in_channels=3,
    out_channels=1,
    padding=0,
    stride=1,
    kernel_size=3
)

output = conv(x)

# print(output.shape)
# print(output)
print(conv.weight.shape)
print(conv.bias.shape)
```
