```python
from pathlib import Path

data_dir = Path("../Data/dataset-resized")

print(data_dir.exists())
```

```python
classes = sorted([folder.name for folder in data_dir.iterdir() if folder.is_dir()])

print(classes)
print("Number of classes:", len(classes))
```

```python
for class_name in classes:
    class_path = data_dir / class_name
    images = list(class_path.glob("*.jpg"))

    print(class_name, ":", len(images))
    
```

```python
from PIL import Image
import matplotlib.pyplot as plt
import random

class_name = random.choice(classes)

class_path = data_dir / class_name
image_path = random.choice(list(class_path.glob("*.jpg")))

image = Image.open(image_path)

print("Class:", class_name)
print("Image size:", image.size)
print("Image mode:", image.mode)

plt.imshow(image)
plt.title(class_name)
plt.axis("off")
plt.show()
```

```python
from pathlib import Path

data_dir = Path("../Data/dataset-resized")

classes = sorted([folder.name for folder in data_dir.iterdir() if folder.is_dir()])

print("Classes:", classes)

for class_name in classes:
    count = len(list((data_dir / class_name).glob("*")))
    print(f"{class_name}: {count}")
```

```python
fig, axes = plt.subplots(2, 3, figsize=(12, 8))

for ax, class_name in zip(axes.ravel(), classes):

    class_path = data_dir / class_name
    image_path = random.choice(list(class_path.glob("*.jpg")))

    image = Image.open(image_path)

    ax.imshow(image)
    ax.set_title(class_name)
    ax.axis("off")

plt.tight_layout()
plt.show()
```
