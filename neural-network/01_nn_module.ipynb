```python
import torch
import torch.nn as nn

class MyModel(nn.Module):

    def __init__(self):
        super().__init__()

        self.w = nn.Parameter(torch.tensor(3.0))

    def forward(self, x):
        return x * self.w

model = MyModel()

x = torch.tensor(2.0)

prediction = model(x)

print(prediction)
```
