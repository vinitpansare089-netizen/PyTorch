```python
import torch
import torch.nn as nn

class MyNetwork(nn.Module):

    def __init__(self):
        super().__init__()

        self.layer1 = nn.Linear(2, 4)
        self.relu1 = nn.ReLU()

        self.layer2 = nn.Linear(4, 3)
        self.relu2 = nn.ReLU()

        self.layer3 = nn.Linear(3, 1)

    def forward(self, x):

        x = self.layer1(x)
        x = self.relu1(x)

        x = self.layer2(x)
        x = self.relu2(x)

        x = self.layer3(x)

        return x

model = MyNetwork()

for name, param in model.named_parameters():
    print(name, param.shape)


x = torch.tensor([
    [1.0, 2.0],
    [2.0, 1.0],
    [3.0, 2.0],
    [4.0, 1.0]
])

output = model(x)

print(output)
print(output.shape)
```
