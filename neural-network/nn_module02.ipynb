```python
import torch
import torch.nn as nn
from torch.utils.data import TensorDataset, DataLoader


# -------------------------
# 1. Data
# -------------------------

x = torch.tensor([
    [1.0, 2.0],
    [2.0, 1.0],
    [3.0, 2.0],
    [4.0, 1.0],
    [5.0, 2.0],
    [6.0, 1.0]
])

y = torch.tensor([
    [8.0],
    [7.0],
    [12.0],
    [11.0],
    [16.0],
    [15.0]
])


# -------------------------
# 2. Dataset + DataLoader
# -------------------------

dataset = TensorDataset(x, y)

loader = DataLoader(
    dataset,
    batch_size=2,
    shuffle=True
)


# -------------------------
# 3. Neural Network
# -------------------------

class MyNetwork(nn.Module):

    def __init__(self):
        super().__init__()

        self.layer1 = nn.Linear(2, 4)
        self.relu1 = nn.ReLU()

        self.layer2 = nn.Linear(4, 3)
        self.relu2 = nn.ReLU()

        self.layer3 = nn.Linear(3, 1)

    def forward(self, x):

        x = self.layer1(x)
        x = self.relu1(x)

        x = self.layer2(x)
        x = self.relu2(x)

        x = self.layer3(x)

        return x


# -------------------------
# 4. Model + Loss + Optimizer
# -------------------------

model = MyNetwork()

loss_fn = nn.MSELoss()

optimizer = torch.optim.SGD(
    model.parameters(),
    lr=0.01
)


# -------------------------
# 5. Training
# -------------------------

for epoch in range(10):

    total_loss = 0

    for batch_x, batch_y in loader:

        prediction = model(batch_x)

        loss = loss_fn(prediction, batch_y)

        loss.backward()

        optimizer.step()

        optimizer.zero_grad()

        total_loss += loss.item()

    average_loss = total_loss / len(loader)

    print("Epoch:", epoch, "Average Loss:", average_loss)
```
