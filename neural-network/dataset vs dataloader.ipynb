```python
"""
Dataset va data loader"""

import torch
from torch.utils.data import TensorDataset, DataLoader


x = torch.tensor([
    [1.0, 2.0],
    [2.0, 1.0],
    [3.0, 2.0],
    [4.0, 1.0],
    [5.0, 2.0],
    [6.0, 1.0]
])

y = torch.tensor([
    [8.0],
    [7.0],
    [12.0],
    [11.0],
    [16.0],
    [15.0]
])


dataset = TensorDataset(x,y)
# print(dataset)

loader = DataLoader(
    dataset,
    batch_size=2,
    shuffle=True
)

for batch_x, batch_y in loader:
    print("X shape:", batch_x.shape)
    print("Y shape:", batch_y.shape)
    print("X:", batch_x)
    print("Y:", batch_y)
    print("-----")
```
