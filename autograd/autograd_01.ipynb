```python
"""
Autograd — Multiple Variables
"""

import torch

x = torch.tensor(2.0, requires_grad=True)
w = torch.tensor(3.0, requires_grad=True)

y = x * w

print("y:", y)

y.backward()

print("gradient of x:", x.grad)
print("gradient of w:", w.grad)
```

```python
"""
Loss Function
"""

prediction = torch.tensor(6.0)
actual = torch.tensor(10.0)

loss = (prediction - actual) ** 2

print("prediction: ", prediction)
print("actual: ", actual)
print("Loss: ", loss)
```

```python
import torch

x = torch.tensor(2.0)
w = torch.tensor(3.0, requires_grad=True)

actual = torch.tensor(10.0)

prediction = x * w

loss = (prediction - actual) ** 2

print("Prediction:", prediction)
print("Loss:", loss)

loss.backward()

print("Gradient of w:", w.grad)
```

```python
import torch
import torch.nn as nn

x = torch.tensor([
    [1.0, 2.0],
    [2.0, 1.0],
    [3.0, 2.0],
    [4.0, 1.0]
])

y = torch.tensor([
    [8.0],
    [7.0],
    [12.0],
    [11.0]
])

model = nn.Sequential(
    nn.Linear(2, 4),
    nn.ReLU(),
    nn.Linear(4, 3),
    nn.ReLU(),
    nn.Linear(3, 1)
)

prediction = model(x)

print(prediction)

loss_fn = nn.MSELoss()

optimizer = torch.optim.SGD(
    model.parameters(),
    lr=0.01
)

loss = loss_fn(prediction, y)
print(loss)

loss.backward()

for name, param in model.named_parameters():
    print(name)
    print(param.grad)

optimizer.step()
print(model[4].bias)
```

```python
for epoch in range(1200):

    # 1. Forward
    prediction = model(x)

    # 2. Loss
    loss = loss_fn(prediction, y)

    # 3. Calculate gradients
    loss.backward()

    # 4. Update weights
    optimizer.step()

    # 5. Clear gradients
    optimizer.zero_grad()

    if epoch % 100 == 0:
        print("Epoch:", epoch, "Loss:", loss.item())

print("Predictions:")
print(model(x))

print("\nActual:")
print(y)
```
