```python
"""
First autograd
"""
import torch

x = torch.tensor(3.0, requires_grad=True)

y = x ** 2

print(x)
print(y)

y.backward()

print(x.grad)
```

```python
import torch

x = torch.tensor(2.0, requires_grad=True)

y = x ** 2 + 3 * x

print('Y: ', y)

y.backward()

print("gradient: ", x.grad)
```
