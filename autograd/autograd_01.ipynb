```python
"""
First autograd
"""
import torch

x = torch.tensor(3.0, requires_grad=True)

y = x ** 2

print(x)
print(y)

y.backward()

print(x.grad)
```

```python
import torch

x = torch.tensor(2.0, requires_grad=True)

y = x ** 2 + 3 * x

print('Y: ', y)

y.backward()

print("gradient: ", x.grad)
```

```python
"""
Autograd — Multiple Variables
"""

import torch

x = torch.tensor(2.0, requires_grad=True)
w = torch.tensor(3.0, requires_grad=True)

y = x * w

print("y:", y)

y.backward()

print("gradient of x:", x.grad)
print("gradient of w:", w.grad)
```

```python
"""
Loss Function
"""

prediction = torch.tensor(6.0)
actual = torch.tensor(10.0)

loss = (prediction - actual) ** 2

print("prediction: ", prediction)
print("actual: ", actual)
print("Loss: ", loss)
```

```python
import torch

x = torch.tensor(2.0)
w = torch.tensor(3.0, requires_grad=True)

actual = torch.tensor(10.0)

prediction = x * w

loss = (prediction - actual) ** 2

print("Prediction:", prediction)
print("Loss:", loss)

loss.backward()

print("Gradient of w:", w.grad)
```

```python
import torch
import torch.nn as nn

model = nn.Sequential(
    nn.Linear(5, 4),
    nn.ReLU(),
    nn.Linear(4, 3),
    nn.ReLU(),
    nn.Linear(3, 1)
)

x = torch.tensor([[2.0, 3.0, 1.0, 4.0, 5.0]])

x1 = model[0](x)
print("After Linear 1:", x1)

x2 = model[1](x1)
print("After ReLU 1:", x2)

x3 = model[2](x2)
print("After Linear 2:", x3)

x4 = model[3](x3)
print("After ReLU 2:", x4)

x5 = model[4](x4)
print("Final output:", x5)
```
