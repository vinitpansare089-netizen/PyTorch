```python
import torch

# Input and correct answer
x = torch.tensor(2.0)
actual = torch.tensor(10.0)

# Starting weight
w = torch.tensor(3.0, requires_grad=True)

# Learning rate
learning_rate = 0.01

for i in range(10):

    # 1. Make prediction
    prediction = x * w

    # 2. Calculate loss
    loss = (prediction - actual) ** 2

    # 3. Calculate gradient
    loss.backward()

    # 4. Update weight
    with torch.no_grad():
        w -= learning_rate * w.grad

    # 5. Clear old gradient
    w.grad.zero_()

    if i % 10 == 0:
        print(
            "Step:", i,
            "Weight:", w.item(),
            "Prediction:", prediction.item(),
            "Loss:", loss.item()
        )
```

```python
"""
Optimiser(SGD) used with learning rate 
"""
import torch

x = torch.tensor(2.0)
actual = torch.tensor(10.0)

w = torch.tensor(3.0, requires_grad=True)

optimizer = torch.optim.SGD([w], lr=0.01)

for i in range(10):

    prediction = x * w

    loss = (prediction - actual) ** 2

    loss.backward()

    optimizer.step()

    optimizer.zero_grad()

    print(
        "Step:", i,
        "Weight:", w.item(),
        "Prediction:", prediction.item(),
        "Loss:", loss.item()
    )
```
