```python
import torch
import torch.nn as nn

model = nn.Linear(1, 1)

with torch.no_grad():
    model.weight.fill_(2.0)
    model.bias.fill_(1.0)

x = torch.tensor([[3.0]])
y = model(x)

print("Input:", x)
print("Weight:", model.weight)
print("Bias:", model.bias)
print("Output:", y.item())
```
