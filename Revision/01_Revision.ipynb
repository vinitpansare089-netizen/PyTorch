```python
import torch 

x = torch.tensor([1,2,3,4])

print(x)
print(x.shape)
print(x.ndim)
print(x.dtype)
```

```python
import torch

x = torch.tensor([
    [1,5],
    [3,4]
])

y = torch.tensor([
    [4,3],
    [1,2]
])

# print(x*y)
print(torch.matmul(x,y))
```

```python
"""
Exercise 2 — 2D Tensor
"""

import torch

x = torch.Tensor([
    [1,2,3],
    [4,5,6],
    [7,8,9]
])

print(x)
print(x.ndim)
print(x.shape)
print(x[2][1])
```

```python
"""
Exercise 3 — Element-wise operations
"""
x = torch.tensor([1,2,3])
y = torch.tensor([4,5,6])

print(x + y)
print(x - y)
print(x * y)
print(x / y)
```

```python
x = torch.tensor([
    [1,2,3],
    [4,5,6]
])

print(x)
print(x.shape)

y = x.reshape(3,2)
print(y.shape)
print(y)

```
