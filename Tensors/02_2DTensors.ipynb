```python
"""
Exercise for 2D tensors
"""
import torch

x = torch.tensor([
    [5, 10, 15],
    [20, 25, 30],
    [35, 40, 45]
])

print(x)
print(x.shape)
print(x.dtype)
print(x.ndim)
print(x[1])
print(x[2][1])

```

```python
"""
tensors operations
"""

import torch 

x = torch.tensor([10,20,30])
y = torch.tensor([1,2,3])

print(x + y)
print(x - y)
print(x * y)
print(x / y)
print((x / y).dtype)
```

```python
"""
Matrix multipliation
"""

import torch

A = torch.tensor([
    [1,2],
    [3,4]
])

B = torch.tensor([
    [5,6],
    [7,8]
])

# print(A * B) /// Element-Wise multiplication

print(torch.matmul(A,B)) #/// Real matrix multiplication
```

```python
"""
Reshaping Tensors
"""
import torch

x = torch.tensor([
    [1,2,3],
    [4,5,6]
])

print(x)
print(x.shape)

y = x.reshape(3,2)
print(y)
print(y.shape)
```
