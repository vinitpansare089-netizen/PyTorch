```python
"""
Exercise for 2D tensors
"""
import torch

x = torch.tensor([
    [5, 10, 15],
    [20, 25, 30],
    [35, 40, 45]
])

print(x)
print(x.shape)
print(x.dtype)
print(x.ndim)
print(x[1])
print(x[2][1])

```

```python
"""
Matrix multiplication
"""

import torch 

x = torch.tensor([10,20,30])
y = torch.tensor([1,2,3])

print(x + y)
print(x - y)
print(x * y)
print(x / y)
```
