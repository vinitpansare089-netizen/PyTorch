```python
"""
tensors operations
"""

import torch 

x = torch.tensor([10,20,30])
y = torch.tensor([1,2,3])

print(x + y)
print(x - y)
print(x * y)
print(x / y)
print((x / y).dtype)
```

```python
"""
Matrix multipliation
"""

import torch

A = torch.tensor([
    [1,2],
    [3,4]
])

B = torch.tensor([
    [5,6],
    [7,8]
])

# print(A * B) /// Element-Wise multiplication

print(torch.matmul(A,B)) #/// Real matrix multiplication
```

```python
"""
dtype and conversions
"""
import torch

x = torch.tensor([1,2,3])
y = x.float()

# print(y)
# print(y.dtype)
# print(x)
print(x.dtype)

x = torch.tensor([1,2,3],  dtype=torch.float32)
print(x)
print(x.dtype)
```

```python
"""
Tensor to numpy
"""

x = torch.tensor([1,2,3])

y = x.numpy()
print(y)
print(y.dtype)
```

```python
"""
numpy to torch
"""
import numpy as np

x = np.array([1,2,3])

y = torch.from_numpy(x)

print(x)
print(x.dtype)
print(type(x))
```
